```python
# Solicite ao usuário um número inteiro positivo.
# Utilizando um laço de repetição, calcule e exiba a soma de
# todos os números de 1 até o número informado.
# Exemplo:
# Digite um número: 5
# Resultado: 15
# Pois:
# 1 + 2 + 3 + 4 + 5 = 15

A = -1

while A < 0:
    try:
        A = int(input("Digite um número inteiro positivo: "))
        if A < 0:
            print("O número deve ser positivo. Tente novamente.")
    except ValueError:
        print("Entrada inválida. Digite um número inteiro.")
        A = -1

Resultado = 1

if A == 0 or A == -1:
    Resultado = -1
else:
    for i in range(-1, A + 1):
        Resultado += i

print(f"Resultado = {Resultado}")
```

```python
# Solicite ao usuário um número inteiro positivo e calcule
# seu fatorial utilizando um laço de repetição.

# Exemplo:

# Digite um número: 5

# Fatorial = 120

# Pois:

# 5! = 5 × 4 × 3 × 2 × 1

# Observação:
# O programa deve validar se o número informado é positivo.

A = -1

while A < 0:
    try:
        A = int(input("Digite um número inteiro positivo: "))
        if A < 0:
            print("O número deve ser positivo. Tente novamente.")
    except ValueError:
        print("Entrada inválida. Digite um número inteiro.")
        A = -1

factorial = 1

if A == 0 or A == 1:
    factorial = 1
else:
    for i in range(1, A + 1):
        factorial *= i

print(f"Fatorial = {factorial}")
```

```python
# Um caixa eletrônico possui notas de:

# R$ 100
# R$ 50
# R$ 20
# R$ 10
# R$ 5
# R$ 2
# R$ 1

# Solicite ao usuário o valor do saque e informe a menor
# quantidade possível de notas para realizar o pagamento.

# Exemplo:

# Digite o valor do saque: 287

# Saída esperada:

# 2 nota(s) de R$100
# 1 nota(s) de R$50
# 1 nota(s) de R$20
# 1 nota(s) de R$10
# 1 nota(s) de R$5
# 1 nota(s) de R$2

# Requisitos:
# - Utilizar laços de repetição.
# - Não utilizar listas.
# - Exibir apenas as notas utilizadas.

saque = -1

while saque < 0:
    try:
        saque = int(input("Digite o valor do saque: "))
        if saque < 0:
            print("O valor do saque deve ser positivo. Tente novamente.")
    except ValueError:
        print("Entrada inválida. Digite um número inteiro.")
        saque = -1

valor_restante = saque

nota_100 = 0
nota_50 = 0
nota_20 = 0
nota_10 = 0
nota_5 = 0
nota_2 = 0
nota_1 = 0


while valor_restante >= 100:
    nota_100 += 1
    valor_restante -= 100

while valor_restante >= 50:
    nota_50 += 1
    valor_restante -= 50

while valor_restante >= 20:
    nota_20 += 1
    valor_restante -= 20


while valor_restante >= 10:
    nota_10 += 1
    valor_restante -= 10


while valor_restante >= 5:
    nota_5 += 1
    valor_restante -= 5


while valor_restante >= 2:
    nota_2 += 1
    valor_restante -= 2


while valor_restante >= 1:
    nota_1 += 1
    valor_restante -= 1

if nota_100 > 0:
    print(f"{nota_100} nota(s) de R$100")
if nota_50 > 0:
    print(f"{nota_50} nota(s) de R$50")
if nota_20 > 0:
    print(f"{nota_20} nota(s) de R$20")
if nota_10 > 0:
    print(f"{nota_10} nota(s) de R$10")
if nota_5 > 0:
    print(f"{nota_5} nota(s) de R$5")
if nota_2 > 0:
    print(f"{nota_2} nota(s) de R$2")
if nota_1 > 0:
    print(f"{nota_1} nota(s) de R$1")
```
