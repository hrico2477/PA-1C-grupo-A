```python
# Exercício 06

# Considere:

notas = [7.5, 8.0, 6.5, 9.0, 5.5]

# Mostre:

# A maior nota
# A menor nota
# A média das notas

notas = [7.5, 8.0, 6.5, 9.0, 5.5]


maior = max(notas)
menor = min(notas)

media = sum(notas) / len(notas)

print("A maior nota :", maior)
print("A menor nota :", menor)
print("A média das notas :", media)
```

```python
# Execício 07
# crie uma lista vazia chamadas produtos
# peça ao usuário  para cadastrar 5 prodtus utilizando for
# mostre os produtos númerados

produtos = []


for i in range(5):
    item = input(f"Cadastre o produto {i+1}: ")
    produtos.append(item)


for indice, produto in enumerate(produtos, start=1):
    print(f"# {indice} - {produto}")
```

```python
# Exercício 08
# considere:
alunos =["Ana","carlos","Maria","Pedro","João"]
# Peça ao usuário para digitar um nome, o progr5ama deverá verficar se o nome costa na lista ou não
# faça o programa funcion indepedente da forma que o usário digita, ex: Marwia, maria ou MARIA

alunos = ["Ana", "Carlos", "Maria", "Pedro", "João"]

alunos_lower = [aluno.lower() for aluno in alunos]



busca = input("Digite o nome do aluno para buscar: ")

if busca.lower() in alunos_lower:
    print(f"O aluno(a) {busca} consta na lista!")
else:
    print(f"O aluno(a) {busca} NÃO consta na lista.")
```

```python
# Exercício 09

# crie uma lista vazia chamada numeros
# peça o usuário para informar 10 números
# depois crie duas listas vazias chamadas pares e impares
# Percorra a lista orginal e coloque cadanúmero na lista correspondete.
numeros = []


for i in range(10):
    num = int(input(f"Digite o {i+1}º número inteiro: "))
    numeros.append(num)


pares = []
impares = []


for num in numeros:
    if num % 2 == 0:
        pares.append(num)
    else:
        impares.append(num)



print("Lista original:", numeros)
print("Números pares:", pares)
print("Números ímpares:", impares)

```
