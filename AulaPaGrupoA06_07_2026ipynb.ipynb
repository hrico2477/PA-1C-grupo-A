```python
meta = 10000.0
total_acumulado = 0.0
meses = 0

print("---Simulador de Cofrinho (Meta: R$ 10.000)---")

while total_acumulado < meta:
    meses += 1
    deposito = float(input(f"Informe o valor do depósito para o mês {meses}: R$ "))
    total_acumulado += deposito
3
print("---Meta Atingida---")
print(f"Quantos meses foram necessários: {meses}")
print(f"O total acumulado: R$ {total_acumulado:.2f}")
print(f"Em qual mês a meta foi atingida: Mês {meses}")
```

```python
subtotal_geral = 0.0

print("--- aixa Registradora---")
print("Digite '0' no preço para encerrar a compra.\n")

while True:
    preco = float(input("Preço do produto: R$ "))

    if preco == 0:
        break

    quantidade = int(input("Quantidade comprada: "))

    subtotal_geral += preco * quantidade
    print("-" * 20)

if subtotal_geral < 100:
    porcentagem_desconto = 0
elif subtotal_geral < 300:
    porcentagem_desconto = 5
elif subtotal_geral < 500:
    porcentagem_desconto = 10
else:
    porcentagem_desconto = 15

valor_desconto = subtotal_geral * (porcentagem_desconto / 100)
valor_final = subtotal_geral - valor_desconto

print("\n==========RESUMO DA COMPRA==========")
print(f"Subtotal: R$ {subtotal_geral:.2f}")
print(f"Desconto aplicado: {porcentagem_desconto}% (Economia de R$ {valor_desconto:.2f})")
print(f"Valor final: R$ {valor_final:.2f}")
print("=======================================")
```
