```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]

soma = sum(notas)
media = soma / len(notas)
maior = max(notas)
menor = min(notas)

print(f"Soma das notas: {soma}")
print(f"Média: {media:.2f}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")


if media >= 7:
    print("Situação: Aprovado")
else:
    print("Situação: Reprovado")
```

```python
numeros = list(range(1, 21))

pares = []
for n in numeros:
    if n % 2 == 0:
        pares.append(n)

print("Pares (for):", pares)

pares_comprehension = [n for n in numeros if n % 2 == 0]
print("Pares (comprehension):", pares_comprehension)
```

```python
numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

positivos = []
negativos = []
zeros = []

for num in numeros:
    if num > 0:
        positivos.append(num)
    elif num < 0:
        negativos.append(num)
    else:
        zeros.append(num)

print("Positivos:", positivos)
print("Negativos:", negativos)
print("Zeros:", zeros)
```

```python
numeros = [15, 3, 89, 42, 7, 21, 64, 5, 90, 11]

print("Lista original:", numeros)

lista_copia = numeros.copy()

crescente = sorted(lista_copia)

decrescente = sorted(lista_copia, reverse=True)

print("Crescente:", crescente)
print("Decrescente:", decrescente)
print("Maior valor:", max(lista_copia))
print("Menor valor:", min(lista_copia))
print("Lista original mantida:", numeros)
```

```python
produtos = [
    ["Arroz", 25.90],
    ["Feijão", 8.50],
    ["Macarrão", 6.90],
    ["Leite", 5.50],
    ["Café", 15.00],
]

print("--- Lista de Produtos ---")
for nome, preco in produtos:
    print(f"{nome}: R$ {preco:.2f}")

valor_total = sum(preco for _, preco in produtos)
produto_mais_caro = max(produtos, key=lambda x: x[1])
produto_mais_barato = min(produtos, key=lambda x: x[1])

print(f"\nValor total: R$ {valor_total:.2f}")
print(
    f"Mais caro: {produto_mais_caro[0]} (R$ {produto_mais_caro[1]:.2f})"
)
print(
    f"Mais barato: {produto_mais_barato[0]} (R$ {produto_mais_barato[1]:.2f})"
)

busca_prod = input("\nDigite o nome de um produto para buscar: ").strip()
encontrado = any(
    p[0].lower() == busca_prod.lower() for p in produtos
)

if encontrado:
    print(f"O produto '{busca_prod}' está cadastrado!")
else:
    print(f"O produto '{busca_prod}' NÃO foi encontrado.")
```

```python
qtd_alunos = int(input("Quantos alunos serão cadastrados? "))
turma = []

for i in range(qtd_alunos):
    print(f"\n--- Aluno {i+1} ---")
    nome = input("Nome: ").strip()
    nota = float(input("Nota: "))
    turma.append([nome, nota])


soma_notas = sum(aluno[1] for aluno in turma)
media_turma = soma_notas / len(turma)

mais_nota = max(turma, key=lambda x: x[1])
menos_nota = min(turma, key=lambda x: x[1])

aprovados = [aluno[0] for aluno in turma if aluno[1] >= 7.0]
reprovados = [aluno[0] for aluno in turma if aluno[1] < 7.0]

print("\n" + "=" * 30)
print("RELATÓRIO DA TURMA")
print("=" * 30)

print("\nAlunos Cadastrados:")
for nome, nota in turma:
    print(f"- {nome}: {nota:.1f}")

print(f"\nMédia da turma: {media_turma:.2f}")
print(f"Maior nota: {mais_nota[0]} ({mais_nota[1]:.1f})")
print(f"Menor nota: {menos_nota[0]} ({menos_nota[1]:.1f})")
print("Aprovados:", ", ".join(aprovados) if aprovados else "Nenhum")
print("Reprovados:", ", ".join(reprovados) if reprovados else "Nenhum")

print("\n" + "=" * 30)
pesquisa = input("Digite o nome de um aluno para pesquisar a nota: ").strip()
encontrado = False

for nome, nota in turma:
    if nome.lower() == pesquisa.lower():
        print(f"A nota de {nome} é: {nota:.1f}")
        encontrado = True
        break

if not encontrado:
    print(f"Aluno '{pesquisa}' não encontrado no cadastro.")
```
