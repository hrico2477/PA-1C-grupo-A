```python
#  Escrever um programa que recebe 4 notas, e calcula a média e imprime se o aluno foi aprovado, recuperação ou reprovado

# 7 ou mais aprovado
# 5 ou mais recuperação
# menor que 5 reprovado
# use while e except ValueError: por favor
notas = []
quantidade_notas = 4


while len(notas) < quantidade_notas:
    try:
        entrada = input(f"Digite a nota {len(notas) + 1}: ")
        nota = float(entrada)


        if 0 <= nota <= 10:
            notas.append(nota)
        else:
            print("Por favor, digite uma nota válida entre 0 e 10.")

    except ValueError:
        print("Entrada inválida! Digite apenas números.")


media = sum(notas) / quantidade_notas
print(f"\nMédia final: {media:.2f}")


if media >= 7:
    print("Situação: APROVADO")
elif media >= 5:
    print("Situação: RECUPERAÇÃO")
else:
    print("Situação: REPROVADO")
```
